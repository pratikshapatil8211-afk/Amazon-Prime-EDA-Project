###3.1 Feature Engineering

#### Objective

The objective of this section is to **create new meaningful features** from the existing data, making the dataset **more informative** and **suitable for deeper analysis**.

````python
# Create a new column to calculate the age of each title
from datetime import datetime

current_year = datetime.now().year

titles['content_age'] = current_year - titles['release_year']
````

````python
# Extract Primary Genre
# Extract the first genre from the list for easier analysis

titles['primary_genre'] = (
    titles['genres']
    .str.strip('[]')
    .str.replace("'", "", regex=False)
    .str.split(',')
    .str[0]
    .str.strip()
)
````

````python
# Extract Primary Production Country
# Extract the first production country from the list

titles['primary_country'] = (
    titles['production_countries']
    .str.strip('[]')
    .str.replace("'", "", regex=False)
    .str.split(',')
    .str[0]
    .str.strip()
)
````

````python
# Verify Newly Created Features
# Display the newly created columns to verify the results

display(titles[['title', 'release_year', 'content_age', 'primary_genre', 'primary_country']].head())
````

````python
# Validate Required Columns

required_columns = [
    'primary_genre',
    'primary_country',
    'content_age'
]

for column in required_columns:
    if column in titles.columns:
        print(f"✅ {column} created successfully.")
    else:
        print(f"❌ {column} is missing.")
````

### Chart 1: Distribution of Movies and TV Shows

This visualization compares the number of **Movies** and **TV Shows** available on Amazon Prime. It provides an overview of the platform's content distribution and helps understand whether Amazon Prime focuses more on movies or television series.

````python
# Chart 1: Distribution of Movies and TV Shows

# Create a figure for visualization
plt.figure(figsize=(8,6))

# Plot the count of Movies and TV Shows
ax = sns.countplot(
    data=titles,
    x='type',
    hue='type',          # Used to avoid future seaborn warning
    palette=custom_palette,
    legend=False
)

# Display the count on top of each bar
for container in ax.containers:
    ax.bar_label(container, fontsize=11)

# Add chart title and axis labels
plt.title(
    "Distribution of Movies and TV Shows on Amazon Prime",
    fontsize=15,
    fontweight='bold'
)

plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

# Add grid lines for better readability
plt.grid(axis='y', linestyle='--', alpha=0.4)

# Adjust layout
plt.tight_layout()

# Display the chart
plt.show()
````

##### 1. Why did you pick the specific chart?

A **Count Plot** was selected because **Content Type** is a **categorical variable**. It is the most suitable chart for comparing the number of **Movies** and **TV Shows** available on Amazon Prime, making the distribution easy to understand.

### Chart 2: Content Released Over the Years

This visualization shows the number of **Movies** and **TV Shows** released over the years on Amazon Prime. It helps identify **release trends**, understand how the content library has grown over time, and highlights periods of significant expansion.

````python
# Chart 2: Content Released Over the Years

# Count the number of titles released each year
release_year = titles['release_year'].value_counts().sort_index()

# Create figure
plt.figure(figsize=(12,6))

# Create line chart
plt.plot(
    release_year.index,
    release_year.values,
    color="#122D70",
    linewidth=2.5,
    marker='o',
    markersize=5
)

# Set background color
plt.gca().set_facecolor("#F4FFFB")
plt.gcf().set_facecolor("#F4FFFB")

# Add chart title and axis labels
plt.title("Content Released Over the Years", fontsize=15, fontweight='bold')
plt.xlabel("Release Year", fontsize=12)
plt.ylabel("Number of Titles", fontsize=12)

# Add grid for better readability
plt.grid(True, linestyle='--', alpha=0.4)

# Remove top and right borders
sns.despine()

# Adjust layout
plt.tight_layout()

# Display the chart
plt.show()
````

##### 1. Why did you pick the specific chart?

A **Line Chart** was chosen because **release_year** is a **time-based variable**. It clearly shows how the number of titles released on Amazon Prime has changed over time, making trends and growth easy to understand.

##### 2. What is/are the insight(s) found from the chart?

The number of titles released on Amazon Prime increased steadily over the years, with rapid growth after **2015**. The highest number of titles (**856**) was released in **2021**, showing a major expansion of the content library. The decline in **2022** is likely due to **incomplete data** for that year.
